### Situación de Interés: La heterogeneidad espacial en la agricultura
En las semanas anteriores analizamos el comportamiento de las variables de forma general. Sin embargo, en la ingeniería agrícola **el campo nunca es homogéneo**.

Un lote de producción está dividido en zonas geográficas y posee diferentes texturas de suelo. Si calculamos el promedio de rendimiento de toda la finca, podríamos ocultar el hecho de que la zona Norte está produciendo el doble que la zona Sur debido al tipo de suelo. Hoy aprenderemos a agrupar los datos (segmentación) para descubrir estas realidades ocultas.

```

### Celda de Código 1

````
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
np.random.seed(42)

# Simulamos 200 registros de parcelas en diferentes zonas y con distintos suelos
zonas = np.random.choice(['Norte', 'Sur', 'Este', 'Oeste'], size=200)
textura = np.random.choice(['Arcilloso', 'Franco', 'Arenoso'], size=200, p=[0.3, 0.5, 0.2])

# La humedad base depende del tipo de suelo
humedad_base = np.where(textura == 'Arcilloso', 30, np.where(textura == 'Franco', 22, 12))
humedad = humedad_base + np.random.normal(loc=0, scale=3, size=200)

# El rendimiento se ve afectado por la humedad y un "bono" si es suelo Franco
rendimiento = 3 + (humedad * 0.15) + np.where(textura == 'Franco', 2.5, 0) + np.random.normal(0, 1.5, 200)

# Construimos el DataFrame
df_lotes = pd.DataFrame({
    'ID_Parcela': range(1, 201),
    'Zona': zonas,
    'Textura_Suelo': textura,
    'Humedad_%': humedad,
    'Rendimiento_ton_ha': rendimiento
})

print("Datos de heterogeneidad espacial cargados.")
df_lotes.head()
````

### 1. Resúmenes Estadísticos Agrupados (`groupby`)

Para saber qué tipo de suelo retiene más humedad o produce más, no nos sirve el promedio total. Necesitamos dividir los datos por categorías. En `pandas`, esto se logra con el poderoso método `.groupby()`.

```

### Celda de Código 2

````
# Agrupamos los datos por Textura de Suelo y calculamos la media de Humedad y Rendimiento
resumen_suelos = df_lotes.groupby('Textura_Suelo')[['Humedad_%', 'Rendimiento_ton_ha']].mean()

print("Promedios según la textura del suelo:")
round(resumen_suelos, 2)
````

### 2. Comparación Visual de Grupos (Boxplots Categóricos)
Las tablas de promedios son útiles, pero un gráfico de cajas (`boxplot`) nos permite comparar visualmente la mediana, los cuartiles y detectar si hay parcelas con un rendimiento atípicamente bajo o alto dentro de cada categoría de suelo.

```

### Celda de Código 3

````
plt.figure(figsize=(8, 5))
sns.boxplot(data=df_lotes, x='Textura_Suelo', y='Rendimiento_ton_ha', palette='Set2')

plt.title('Variabilidad del Rendimiento según el Tipo de Suelo')
plt.xlabel('Textura del Suelo')
plt.ylabel('Rendimiento (ton/ha)')
plt.show()
````

### 3. Análisis Multivariado con Facetas (Trellis Displays)

¿Qué pasa si queremos cruzar TRES variables a la vez? Por ejemplo, queremos ver la relación entre Humedad y Rendimiento, pero separada por Zona geográfica.

Usaremos `sns.relplot` (o `FacetGrid`) para crear una "cuadrícula" de gráficos automáticamente. Esto es fundamental para identificar si un patrón de ingeniería se mantiene constante en todo el espacio o si cambia según el sector.

```

### Celda de Código 4

````
# Creamos un gráfico de dispersión segmentado por Zona (facetas) y coloreado por Textura
g = sns.relplot(
    data=df_lotes,
    x='Humedad_%',
    y='Rendimiento_ton_ha',
    hue='Textura_Suelo',
    col='Zona',      # Crea un gráfico diferente por cada zona
    col_wrap=2,      # Máximo 2 gráficos por fila
    palette='muted',
    height=3.5,
    aspect=1.2
)

g.fig.suptitle('Relación Humedad-Rendimiento por Zona y Tipo de Suelo', y=1.03)
plt.show()
````
