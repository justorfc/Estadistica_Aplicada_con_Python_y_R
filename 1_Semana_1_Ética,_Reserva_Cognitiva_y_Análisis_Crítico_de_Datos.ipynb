### Situación de Interés: El peligro de la confianza ciega
Un sistema automatizado de Inteligencia Artificial en un cultivo de maíz está recomendando aplicar una lámina de riego de 15 mm. Esta IA toma sus decisiones basándose en un conjunto de sensores de humedad del suelo.

Como ingenieros agrícolas, antes de aprobar esta recomendación (y gastar agua y energía), debemos aplicar nuestro **control humano sobre la IA (human-in-the-loop)**. Nuestra tarea hoy es auditar la integridad de los datos que la máquina está utilizando para tomar esa decisión.

```
# Importamos las librerías necesarias
import pandas as pd
import numpy as np

# Generamos un conjunto de datos simulado de los sensores (Para fines del ejercicio)
# Crearemos intencionalmente un dataset con errores comunes en sensores reales.
datos_simulados = {
    'id_sensor': ['S-01', 'S-02', 'S-03', 'S-04', 'S-05', 'S-06', 'S-07', 'S-08', 'S-09', 'S-10'],
    'zona_cultivo': ['Norte', 'Norte', 'Sur', 'Sur', 'Norte', 'Este', 'Oeste', 'Este', 'Sur', 'Norte'],
    'humedad_volumetrica_porcentaje': [22.5, 21.0, np.nan, -5.0, 23.1, np.nan, 150.5, 20.2, 19.8, 22.0],
    'temperatura_suelo_c': [24.1, 24.5, 23.8, np.nan, 24.2, 25.1, 99.9, 24.0, 23.9, 24.4],
    'estado_bateria': ['OK', 'OK', 'BAJA', 'OK', 'OK', 'CRITICA', 'OK', 'OK', 'OK', 'OK']
}

# Convertimos el diccionario en un DataFrame de pandas
df = pd.DataFrame(datos_simulados)

print("Datos cargados exitosamente en la memoria del cuaderno.")
```

### 2. Estructura y tipos de datos

Ahora necesitamos saber cuánta información tenemos y si Python está interpretando correctamente los números como números y el texto como texto. El método `.info()` es nuestra radiografía.

```
# Consultamos la información técnica del dataset
df.info()
```

### 3. Detección de datos faltantes (NaN)

En el mundo real, los sensores fallan, pierden conexión o se quedan sin batería. Si una IA promedia datos ignorando que hay "huecos", el resultado será erróneo.

Vamos a cuantificar cuántos datos nos faltan en cada variable usando `.isna().sum()`.

```
# Contamos los valores nulos o faltantes por cada columna
df.isna().sum()
```

### 4. Detección de valores atípicos absurdos

Un sensor puede reportar un dato, pero este dato puede ser físicamente imposible. Por ejemplo, ¿puede la humedad volumétrica ser negativa? ¿Puede ser mayor al 100%? ¿Puede el suelo estar a 99.9 °C sin que el cultivo se incinere?

Utilizamos `.describe()` para obtener un resumen estadístico y encontrar anomalías en los valores mínimos y máximos.

```
# Resumen estadístico de las variables numéricas
df.describe()
```
