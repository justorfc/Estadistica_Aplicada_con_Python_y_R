### Situación de Interés: Diseño Hidrológico y Lluvias Extremas
Para diseñar una alcantarilla o un canal de drenaje agrícola, necesitamos conocer la **Precipitación Máxima en 24 horas**. Si subestimamos este valor, la obra se inundará y colapsará.

No nos sirve la distribución Normal porque los eventos extremos (huracanes, tormentas severas) tienen "colas pesadas" que la Normal no alcanza a predecir. En hidrología usamos distribuciones extremas como la **Gumbel** o la **Lognormal**. Hoy tomaremos datos reales históricos, obligaremos a la máquina a buscar los mejores parámetros (ajuste) y verificaremos visual y matemáticamente si el modelo es confiable.

```

### Celda de Código 1

````
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats

sns.set_theme(style="whitegrid")
np.random.seed(42)

# Simulamos 50 años de datos de Precipitación Máxima Anual en 24h (mm)
# Usaremos una distribución Gumbel (asimetría positiva típica de eventos extremos)
precipitacion_maxima = stats.gumbel_r.rvs(loc=80, scale=15, size=50)

print(f"Registro histórico generado: 50 años.")
print(f"Lluvia máxima histórica registrada: {precipitacion_maxima.max():.1f} mm")
print(f"Lluvia mínima histórica registrada: {precipitacion_maxima.min():.1f} mm")
````

### 1. Estimación de Parámetros (Ajuste)

Vamos a "forzar" a la distribución **Normal** y a la distribución **Gumbel** a que se ajusten a nuestros datos empíricos. Para esto, Python utiliza métodos como la Máxima Verosimilitud internamente con la función `.fit()`.

El objetivo es extraer los parámetros matemáticos que definen estas curvas.

```

### Celda de Código 2

````
# 1. Ajustamos a una distribución Normal
# Nos devuelve la media (loc) y la desviación estándar (scale)
mu_norm, std_norm = stats.norm.fit(precipitacion_maxima)

# 2. Ajustamos a una distribución Gumbel
# Nos devuelve el parámetro de ubicación (loc) y escala (scale)
loc_gumbel, scale_gumbel = stats.gumbel_r.fit(precipitacion_maxima)

print(f"Parámetros óptimos NORMAL -> Media: {mu_norm:.2f}, Desv. Estándar: {std_norm:.2f}")
print(f"Parámetros óptimos GUMBEL -> Ubicación: {loc_gumbel:.2f}, Escala: {scale_gumbel:.2f}")

# Visualización inicial con histogramas
plt.figure(figsize=(8, 4))
sns.histplot(precipitacion_maxima, stat='density', color='lightgrey', label='Datos Históricos')

x_rango = np.linspace(40, 160, 100)
plt.plot(x_rango, stats.norm.pdf(x_rango, mu_norm, std_norm), 'r--', lw=2, label='Ajuste Normal')
plt.plot(x_rango, stats.gumbel_r.pdf(x_rango, loc_gumbel, scale_gumbel), 'g-', lw=2.5, label='Ajuste Gumbel')

plt.title('Comparación Visual: Histogramas vs PDFs')
plt.xlabel('Precipitación Máxima en 24h (mm)')
plt.ylabel('Densidad')
plt.legend()
plt.show()
````

### 2. Diagnóstico Visual Avanzado: El Q-Q Plot

Los histogramas engañan. Si cambias el número de barras, la forma cambia.

El **Gráfico Cuantil-Cuantil (Q-Q Plot)** es la prueba visual definitiva. Compara los cuantiles de nuestros datos reales contra los cuantiles teóricos de la distribución. **Si el modelo es perfecto, todos los puntos caerán exactamente sobre una línea recta roja a 45 grados.**

```

### Celda de Código 3

````
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Q-Q Plot para la Distribución Normal
stats.probplot(precipitacion_maxima, dist="norm", sparams=(mu_norm, std_norm), plot=axes[0])
axes[0].set_title('Q-Q Plot: Ajuste Normal')
axes[0].get_lines()[0].set_markerfacecolor('salmon')
axes[0].get_lines()[0].set_markeredgecolor('salmon')

# Q-Q Plot para la Distribución Gumbel
stats.probplot(precipitacion_maxima, dist="gumbel_r", sparams=(loc_gumbel, scale_gumbel), plot=axes[1])
axes[1].set_title('Q-Q Plot: Ajuste Gumbel')
axes[1].get_lines()[0].set_markerfacecolor('mediumseagreen')
axes[1].get_lines()[0].set_markeredgecolor('mediumseagreen')

plt.tight_layout()
plt.show()
````

### 3. Pruebas de Bondad de Ajuste (Prueba de Kolmogorov-Smirnov)

Para tomar decisiones formales, necesitamos matemáticas, no solo apreciaciones visuales.
La prueba de **Kolmogorov-Smirnov (KS)** evalúa la distancia máxima entre la curva acumulada empírica y la teórica.

*   **Hipótesis Nula ($H_0$):** Los datos *sí* siguen la distribución teórica.
*   **Regla de decisión:** Si el valor $p$ ($p$-value) es menor a $0.05$, rechazamos $H_0$ (el ajuste es malo). Si es mayor a $0.05$, no hay evidencia para rechazar el modelo (el ajuste es aceptable).

```

### Celda de Código 4

````
# Prueba KS para la Normal
stat_norm, p_norm = stats.kstest(precipitacion_maxima, 'norm', args=(mu_norm, std_norm))

# Prueba KS para la Gumbel
stat_gumbel, p_gumbel = stats.kstest(precipitacion_maxima, 'gumbel_r', args=(loc_gumbel, scale_gumbel))

print("--- Prueba de Kolmogorov-Smirnov (KS) ---")
print(f"Ajuste NORMAL -> Estadístico KS: {stat_norm:.3f}, p-value: {p_norm:.4f}")
print(f"Ajuste GUMBEL -> Estadístico KS: {stat_gumbel:.3f}, p-value: {p_gumbel:.4f}")

# Criterio automático de ingeniería
print("\nConclusión Técnica:")
if p_norm > p_gumbel:
    print("El modelo NORMAL presenta un mejor ajuste estadístico para este conjunto de datos.")
else:
    print("El modelo GUMBEL presenta un mejor ajuste estadístico para este conjunto de datos, ideal para extremos.")
````
