### Celda de Código 1

````
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
# Fijamos una semilla para que todos en clase obtengamos los mismos resultados iniciales
np.random.seed(42)

print("Entorno de simulación inicializado.")
````

### 1. Probabilidad Simple y Simulación (Ley de los Grandes Números)
Supongamos que los registros históricos de Sincelejo en un mes específico indican que la probabilidad de que un día sea "Lluvioso" es del 30% (0.3) y "Seco" del 70% (0.7).

Vamos a simular esto usando la función `np.random.choice`. Primero simularemos 10 días, luego 100 y finalmente 10,000 para ver cómo se comporta la proporción de días lluviosos.

```

### Celda de Código 2

````
# Definimos nuestro espacio muestral y sus probabilidades teóricas
estados_clima = ['Lluvioso', 'Seco']
probabilidades_teoricas = [0.3, 0.7]

def simular_clima(dias):
    """Simula una cantidad de días y retorna el porcentaje de días lluviosos."""
    simulacion = np.random.choice(estados_clima, size=dias, p=probabilidades_teoricas)
    dias_lluviosos = np.sum(simulacion == 'Lluvioso')
    prob_empirica = dias_lluviosos / dias
    return prob_empirica

# Simulamos diferentes horizontes de tiempo
print(f"Probabilidad empírica en 10 días: {simular_clima(10):.3f}")
print(f"Probabilidad empírica en 100 días: {simular_clima(100):.3f}")
print(f"Probabilidad empírica en 10,000 días: {simular_clima(10000):.3f}")
````

### 2. Eventos Conjuntos e Independencia

Imagina un sistema productivo que depende de dos factores:
1. Que el día sea Seco (70% de probabilidad).
2. Que la bomba de riego falle por problemas eléctricos (10% de probabilidad histórica).

Si asumimos que el clima y la red eléctrica son **independientes**, la probabilidad teórica de que ambas cosas pasen a la vez (Día Seco AND Bomba Falla) es la multiplicación: 0.70 * 0.10 = 0.07 (7%).

¡Vamos a simular 1,000 días para comprobarlo!

```

### Celda de Código 3

````
dias_simulados = 1000

# Simulamos el clima
clima = np.random.choice(['Lluvioso', 'Seco'], size=dias_simulados, p=[0.3, 0.7])

# Simulamos el estado de la bomba
estado_bomba = np.random.choice(['Falla', 'Funciona'], size=dias_simulados, p=[0.1, 0.9])

# Creamos un DataFrame con los resultados
df_eventos = pd.DataFrame({'Clima': clima, 'Bomba': estado_bomba})

# Calculamos cuántas veces ocurrieron ambos eventos simultáneamente
desastre = df_eventos[(df_eventos['Clima'] == 'Seco') & (df_eventos['Bomba'] == 'Falla')]
prob_desastre_empirica = len(desastre) / dias_simulados

print(f"En {dias_simulados} días simulados, tuvimos un Día Seco con Falla de Bomba en {len(desastre)} ocasiones.")
print(f"Probabilidad empírica conjunta: {prob_desastre_empirica:.3f} (Teórica: 0.070)")
````

### 3. Probabilidad Condicional y Tablas de Contingencia

En la realidad, muchas variables **NO** son independientes.
Por ejemplo, si el día es Lluvioso, la probabilidad de que el cultivo sufra "Estrés Hídrico" es bajísima. Pero si el día es Seco, la probabilidad de estrés se dispara, a menos que el riego compense.

Vamos a generar una tabla cruzada (Crosstab) a partir de datos simulados para calcular probabilidades condicionales: **P(Estrés | Día Seco)**.

```

### Celda de Código 4

````
# Simulamos el estado hídrico del cultivo basado en el clima simulado anteriormente
# Si llueve, 95% de las veces está 'Normal' y 5% 'Estrés'.
# Si está seco, 40% 'Normal' y 60% 'Estrés'.

estado_cultivo = []
for c in df_eventos['Clima']:
    if c == 'Lluvioso':
        estado_cultivo.append(np.random.choice(['Normal', 'Estrés'], p=[0.95, 0.05]))
    else:
        estado_cultivo.append(np.random.choice(['Normal', 'Estrés'], p=[0.40, 0.60]))

df_eventos['Estado_Cultivo'] = estado_cultivo

# Generamos la tabla de contingencia (Crosstab) con márgenes
tabla = pd.crosstab(df_eventos['Clima'], df_eventos['Estado_Cultivo'], margins=True)
print("Tabla de Frecuencias Absolutas:")
display(tabla)

# Convertimos a probabilidades condicionales por fila (normalizando el índice)
tabla_condicional = pd.crosstab(df_eventos['Clima'], df_eventos['Estado_Cultivo'], normalize='index')
print("\nProbabilidades Condicionales P(Estado_Cultivo | Clima):")
display(round(tabla_condicional, 3))
````
