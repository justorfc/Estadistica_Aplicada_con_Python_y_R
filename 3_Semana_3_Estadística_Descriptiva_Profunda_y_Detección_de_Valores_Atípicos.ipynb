### Situación de Interés: Variabilidad espacial del suelo
En un proyecto de adecuación de tierras, hemos recolectado muestras de suelo midiendo el porcentaje de Materia Orgánica (MO) y el pH. Sabemos que el promedio nos da una idea general, pero en agricultura de precisión, **la variabilidad es tan importante como el promedio**.

Además, los sensores en campo a veces fallan y registran datos imposibles (valores atípicos o *outliers*). Hoy aprenderemos a medir qué tan dispersos están nuestros datos, hacia dónde se inclinan (sesgo), qué tan concentrados están (curtosis) y cómo aislar matemáticamente los datos anómalos usando Python.

```

### Celda de Código 1

````
# Importamos las librerías
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import skew, kurtosis

sns.set_theme(style="ticks")

# Generamos datos simulados de un lote agrícola (100 muestras)
np.random.seed(123)

# Materia orgánica con una distribución sesgada (log-normal) típica en suelos
materia_organica = np.random.lognormal(mean=1.2, sigma=0.4, size=100)

# pH con distribución normal, pero le inyectaremos valores atípicos artificiales
ph_suelo = np.random.normal(loc=6.5, scale=0.3, size=100)
ph_suelo[10] = 3.1  # Atípico bajo (Error de sensor o zona muy ácida localizada)
ph_suelo[45] = 9.8  # Atípico alto (Derrame de cal agrícola o error)

# Creamos el DataFrame
df_suelos = pd.DataFrame({
    'Muestra_ID': range(1, 101),
    'Materia_Organica_%': materia_organica,
    'pH': ph_suelo
})

print("Datos de suelos cargados. Dimensión:", df_suelos.shape)
df_suelos.head()
````

### 1. Medidas de Forma: Sesgo (Asimetría) y Curtosis

El promedio y la mediana no cuentan toda la historia.
*   **Sesgo (Skewness):** Indica si los datos se alargan hacia la derecha (> 0) o hacia la izquierda (< 0).
*   **Curtosis (Kurtosis):** Indica qué tan pesadas son las colas de la distribución (presencia de valores extremos) comparadas con una campana de Gauss normal (Fisher = 0).

```

### Celda de Código 2

````
# Calculamos Sesgo y Curtosis para la Materia Orgánica
sesgo_mo = df_suelos['Materia_Organica_%'].skew()
curtosis_mo = df_suelos['Materia_Organica_%'].kurtosis() # Por defecto usa la definición de Fisher

print(f"Materia Orgánica - Sesgo: {sesgo_mo:.2f} (Positivo = Cola a la derecha)")
print(f"Materia Orgánica - Curtosis: {curtosis_mo:.2f}")

# Visualizamos para confirmar los números
plt.figure(figsize=(7, 4))
sns.histplot(df_suelos['Materia_Organica_%'], kde=True, color='saddlebrown')
plt.title('Distribución de Materia Orgánica en el Lote')
plt.xlabel('Materia Orgánica (%)')
plt.ylabel('Frecuencia')
plt.axvline(df_suelos['Materia_Organica_%'].mean(), color='red', linestyle='--', label='Media')
plt.axvline(df_suelos['Materia_Organica_%'].median(), color='gold', linestyle='-', label='Mediana')
plt.legend()
plt.show()
````

### 2. Detección de Valores Atípicos (Outliers) en el pH

A simple vista, el promedio puede verse afectado por errores. Vamos a utilizar el **Método del Rango Intercuartílico (IQR)** que calcularon a mano al inicio de la clase, para identificar matemáticamente qué muestras de pH son anormales.

Fórmulas matemáticas:
*   IQR = Q3 - Q1
*   Límite Inferior = Q1 - 1.5 * IQR
*   Límite Superior = Q3 + 1.5 * IQR

```

### Celda de Código 3

````
# 1. Calculamos los cuartiles Q1 (25%) y Q3 (75%)
Q1 = df_suelos['pH'].quantile(0.25)
Q3 = df_suelos['pH'].quantile(0.75)

# 2. Calculamos el IQR
IQR = Q3 - Q1

# 3. Definimos los límites permitidos
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

print(f"Límites normales de pH: Entre {limite_inferior:.2f} y {limite_superior:.2f}")

# 4. Filtramos los datos atípicos
atipicos_ph = df_suelos[(df_suelos['pH'] < limite_inferior) | (df_suelos['pH'] > limite_superior)]

print("\n🚨 Valores Atípicos Detectados en pH:")
atipicos_ph
````

### 3. Visualización de los atípicos

El diagrama de caja (Boxplot) es la representación gráfica exacta de la matemática que acabamos de programar. Los puntos fuera de los "bigotes" son los valores atípicos que el IQR acaba de aislar.

````
plt.figure(figsize=(8, 3))
sns.boxplot(x=df_suelos['pH'], color='lightblue')
plt.title('Boxplot de pH (Puntos aislados = Atípicos)')
plt.xlabel('Nivel de pH')

# Agregamos líneas rojas para marcar nuestros límites matemáticos calculados
plt.axvline(limite_inferior, color='red', linestyle='--', label='Límite Inferior')
plt.axvline(limite_superior, color='red', linestyle='--', label='Límite Superior')
plt.legend()
plt.show()
````
