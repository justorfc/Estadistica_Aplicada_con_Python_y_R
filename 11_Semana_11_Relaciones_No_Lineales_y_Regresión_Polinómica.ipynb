### Situación de Interés: La Ley de Rendimientos Decrecientes
La naturaleza rara vez obedece a una línea recta infinita. En la agronomía existe un principio clave: **la ley de los rendimientos decrecientes**.

Si a un cultivo de arroz le aplicamos Nitrógeno, el rendimiento aumentará drásticamente al principio. Sin embargo, llegará un punto de saturación donde aplicar más Nitrógeno no aumentará la producción. Si seguimos aplicando, entraremos en la fase de toxicidad y el rendimiento caerá.
Una línea recta ($Y = \beta_0 + \beta_1X$) no puede modelar esto. Hoy aprenderemos a curvar nuestra matemática usando **Regresión Polinómica**.

```

### Celda de Código 1

````
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.formula.api as smf

sns.set_theme(style="ticks")
np.random.seed(123)

# Simulamos un ensayo de fertilización con 50 parcelas
# Dosis de Nitrógeno entre 0 y 250 kg/ha
dosis_n = np.random.uniform(0, 250, 50)

# El rendimiento verdadero sigue una curva parabólica (polinomio de grado 2)
# Rendimiento = Base + Efecto Positivo (N) - Efecto Tóxico (N^2)
rendimiento = 2.5 + (0.08 * dosis_n) - (0.0003 * (dosis_n**2)) + np.random.normal(0, 0.6, 50)

df_ensayo = pd.DataFrame({
    'Dosis_N_kgha': dosis_n,
    'Rendimiento_ton': rendimiento
})

print("Datos del ensayo de fertilización generados.")
df_ensayo.head()
````

### 1. El Fracaso del Modelo Lineal Simple
Primero, veamos qué ocurre si intentamos forzar una línea recta a un fenómeno curvo. Ajustaremos el modelo de la semana 9 y veremos el gráfico.

```

### Celda de Código 2

````
# Ajuste Lineal Clásico
modelo_lineal = smf.ols('Rendimiento_ton ~ Dosis_N_kgha', data=df_ensayo).fit()

# Predicciones para graficar una línea suave
x_rango = pd.DataFrame({'Dosis_N_kgha': np.linspace(0, 250, 100)})
y_lineal = modelo_lineal.predict(x_rango)

# Gráfico
plt.figure(figsize=(7, 5))
sns.scatterplot(data=df_ensayo, x='Dosis_N_kgha', y='Rendimiento_ton', color='black', alpha=0.6)
plt.plot(x_rango['Dosis_N_kgha'], y_lineal, color='red', lw=2, label='Ajuste Lineal (Falla)')

plt.title('Ensayo de Fertilización: El error de asumir linealidad')
plt.xlabel('Dosis de Nitrógeno (kg/ha)')
plt.ylabel('Rendimiento (ton/ha)')
plt.legend()
plt.show()

print(f"R-cuadrado del modelo lineal: {modelo_lineal.rsquared:.3f}")
````

### 2. El Rescate Matemático: Regresión Polinómica (Cuadrática)
El gráfico nos grita que necesitamos una curva convexa.
En `statsmodels` (y en R), podemos agregar el término al cuadrado directamente en la fórmula matemática. En lugar de predecir con $X$, predeciremos usando $X$ y $X^2$.

Fórmula estadística: `Y ~ X + I(X**2)`

```

### Celda de Código 3

````
# Ajuste Polinómico de Segundo Grado (Cuadrático)
# Usamos I() para indicarle al modelo que realice la operación matemática de elevar al cuadrado
modelo_cuadratico = smf.ols('Rendimiento_ton ~ Dosis_N_kgha + I(Dosis_N_kgha**2)', data=df_ensayo).fit()

# Extraemos el resumen para ver si el término al cuadrado es estadísticamente significativo (P-value < 0.05)
print(modelo_cuadratico.summary())
````

### 3. Visualización y Determinación del Punto Óptimo
Vamos a dibujar nuestra nueva ecuación curva sobre los datos.
Además, como ingenieros no solo queremos la curva, queremos responder a la pregunta de oro: **¿Cuál es la dosis exacta de Nitrógeno donde el rendimiento llega a su punto máximo?**
Matemáticamente, es el vértice de la parábola: $X_{opt} = \frac{-\beta_1}{2\beta_2}$.

```

### Celda de Código 4

````
# Predicciones de la curva
y_cuadratica = modelo_cuadratico.predict(x_rango)

# Extraemos los coeficientes (betas)
beta_0 = modelo_cuadratico.params['Intercept']
beta_1 = modelo_cuadratico.params['Dosis_N_kgha']
beta_2 = modelo_cuadratico.params['I(Dosis_N_kgha ** 2)']

# Calculamos el vértice (Dosis Óptima)
dosis_optima = -beta_1 / (2 * beta_2)
rendimiento_maximo = beta_0 + (beta_1 * dosis_optima) + (beta_2 * dosis_optima**2)

print(f"🚜 Dosis Óptima calculada matemáticamente: {dosis_optima:.1f} kg/ha")
print(f"🌾 Rendimiento Máximo esperado: {rendimiento_maximo:.2f} ton/ha\n")

# Gráfico del Modelo Cuadrático
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df_ensayo, x='Dosis_N_kgha', y='Rendimiento_ton', color='black', alpha=0.6)
plt.plot(x_rango['Dosis_N_kgha'], y_cuadratica, color='green', lw=3, label='Ajuste Cuadrático')

# Dibujamos el punto óptimo
plt.axvline(dosis_optima, color='green', linestyle='--', alpha=0.5)
plt.plot(dosis_optima, rendimiento_maximo, marker='^', color='gold', markersize=15, markeredgecolor='black', label='Punto Óptimo')

plt.title('Respuesta del Cultivo al Nitrógeno')
plt.xlabel('Dosis de Nitrógeno (kg/ha)')
plt.ylabel('Rendimiento (ton/ha)')
plt.legend()
plt.show()
````
