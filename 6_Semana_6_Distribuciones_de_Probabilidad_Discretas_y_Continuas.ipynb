### Celda de Código 1

````
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats # Librería maestra para distribuciones teóricas en Python

sns.set_theme(style="whitegrid")
np.random.seed(123)

print("Librería scipy.stats cargada. Entorno listo.")
````

### 1. Distribuciones Discretas: Modelando Eventos (Poisson)

Supongamos que en la universidad, una bomba de agua del sistema de riego falla, en promedio, $\lambda = 2$ veces por semestre. ¿Cuál es la probabilidad de que este semestre falle 0 veces? ¿O que falle 5 veces?

Usaremos la distribución de Poisson para simular este escenario y graficar su Función de Masa de Probabilidad (PMF).

```

### Celda de Código 2

````
# Parámetro lambda (tasa promedio de ocurrencia)
tasa_fallas = 2

# Valores posibles (k): desde 0 fallas hasta 8 fallas
k_fallas = np.arange(0, 9)

# Calculamos la probabilidad teórica exacta para cada valor de k usando stats.poisson.pmf
probabilidades_poisson = stats.poisson.pmf(k_fallas, mu=tasa_fallas)

# Visualización
plt.figure(figsize=(7, 4))
plt.bar(k_fallas, probabilidades_poisson, color='steelblue', alpha=0.8)
plt.title(f'Distribución de Poisson ($\lambda$ = {tasa_fallas} fallas por semestre)')
plt.xlabel('Número exacto de fallas en un semestre')
plt.ylabel('Probabilidad Teórica')

# Agregamos las etiquetas de texto sobre cada barra
for i, prob in enumerate(probabilidades_poisson):
    plt.text(i, prob + 0.01, f'{prob:.2f}', ha='center', fontsize=9)

plt.ylim(0, max(probabilidades_poisson) + 0.05)
plt.show()
````

### 2. Distribuciones Continuas: Modelando Fenómenos Físicos

Ahora analicemos la **Precipitación Mensual**. La lluvia no se cuenta (1, 2, 3...), se mide (25.4 mm, 30.1 mm...). Por tanto, es una variable continua.

Como la lluvia mensual no puede ser negativa y muchos meses llueve poco pero unos pocos meses llueve muchísimo (sesgo a la derecha), la **Distribución Normal** suele fallar aquí. La **Distribución Gamma** es una plantilla matemática mucho más adecuada.

Vamos a simular 1000 años de lluvias mensuales usando ambas distribuciones para ver la diferencia visual.

```

### Celda de Código 3

````
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# --- CASO 1: Ajuste Incorrecto (Distribución Normal) ---
# Simulamos lluvia con media = 50mm y desviación estándar = 30mm
lluvia_normal = stats.norm.rvs(loc=50, scale=30, size=1000)

sns.histplot(lluvia_normal, stat='density', ax=axes[0], color='salmon', alpha=0.6)
# Dibujamos la curva teórica (Función de Densidad de Probabilidad - PDF)
x_norm = np.linspace(-50, 150, 100)
axes[0].plot(x_norm, stats.norm.pdf(x_norm, 50, 30), 'r-', lw=2)
axes[0].set_title('Modelo Normal (¡Problema: predice lluvia negativa!)')
axes[0].set_xlabel('Precipitación (mm)')

# --- CASO 2: Ajuste Adecuado (Distribución Gamma) ---
# Usamos un parámetro de forma (a) y escala (scale) adecuados
lluvia_gamma = stats.gamma.rvs(a=2.5, scale=20, size=1000)

sns.histplot(lluvia_gamma, stat='density', ax=axes[1], color='mediumseagreen', alpha=0.6)
x_gam = np.linspace(0, 200, 100)
axes[1].plot(x_gam, stats.gamma.pdf(x_gam, a=2.5, scale=20), 'g-', lw=2)
axes[1].set_title('Modelo Gamma (Solo valores positivos y asimétrica)')
axes[1].set_xlabel('Precipitación (mm)')

plt.tight_layout()
plt.show()
````

### 3. La Función Acumulada (CDF) y los Cuantiles

Como ingenieros, rara vez nos preguntan "¿Cuál es la probabilidad de que llueva *exactamente* 50.12 mm?". Nos preguntan: **"¿Cuál es la probabilidad de que llueva *menos* de 100 mm?"** o **"¿Qué valor de lluvia mensual se supera solo el 10% de los años?"**.

Para responder, usamos:
*   **CDF (Cumulative Distribution Function):** Calcula el área acumulada de izquierda a derecha.
*   **PPF (Percent Point Function) o Cuantiles:** Es la operación inversa. Le damos un porcentaje (ej. 90%) y nos devuelve el valor físico (mm).

```

### Celda de Código 4

````
# Definimos nuestra distribución teórica Gamma para un mes lluvioso
modelo_lluvia = stats.gamma(a=2.5, scale=20)

# Pregunta 1: ¿Cuál es la probabilidad de que llueva MENOS de 100 mm?
prob_menos_100 = modelo_lluvia.cdf(100)
print(f"Probabilidad de precipitación menor a 100 mm: {prob_menos_100:.3f} (o {prob_menos_100*100:.1f}%)")

# Pregunta 2: ¿Cuál es el valor que marca el 90% de los meses? (El cuantil 0.90)
# Es decir, el 90% de las veces lloverá menos que esta cantidad, y solo 10% de las veces lloverá más.
cuantil_90 = modelo_lluvia.ppf(0.90)
print(f"El 90% de las veces, la lluvia mensual será inferior a: {cuantil_90:.1f} mm")
````
